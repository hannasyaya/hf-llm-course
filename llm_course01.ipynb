```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")
classifier(["hello everyone", "I think The Office TV Show is overrated. The jokes from actors are not funny to me"])
```

```python
classifier = pipeline("zero-shot-classification")
classifier(
    """
    CBC News has listened to courtroom submissions in the case, which cast light on the Saanich Police Department investigation. Those records, along with victim impact statements and interviews with victims, raise questions about why police did not act to stop the voyeur sooner.
Saanich police did not answer questions from CBC News for this story, but said in a statement that the department's decisions took a trauma-informed approach to avoid "actions that could contribute to further victimization.
""",
    candidate_labels=["Eduation", "Health", "History"]
)
```
